```python
def generate_chart_code(instruction: str, model: str, out_path_v1: str) -> str:
    """Generate Python code to make a plot with matplotlib using tag-based wrapping."""

    prompt = f"""
    You are a data visualization expert.

    Return your answer *strictly* in this format:

    <execute_python>
    # valid python code here
    </execute_python>

    Do not add explanations, only the tags and the code.

    The code should create a visualization from a DataFrame 'df' with these columns:
    - date (M/D/YY)
    - time (HH:MM)
    - cash_type (card or cash)
    - card (string)
    - price (number)
    - coffee_name (string)
    - quarter (1-4)
    - month (1-12)
    - year (YYYY)

    User instruction: {instruction}

    Requirements for the code:
    1. Assume the DataFrame is already loaded as 'df'.
    2. Use matplotlib for plotting.
    3. Add clear title, axis labels, and legend if needed.
    4. Save the figure as '{out_path_v1}' with dpi=300.
    5. Do not call plt.show().
    6. Close all plots with plt.close().
    7. Add all necessary import python statements

    Return ONLY the code wrapped in <execute_python> tags.
    """

    response = utils.get_response(model, prompt)
    return response
```

```python
# Generate initial code
code_v1 = generate_chart_code(
    instruction="Create a plot comparing Q1 coffee sales in 2024 and 2025 using the data in coffee_sales.csv.", 
    model="openai/gpt-3.5-turbo", 
    out_path_v1="chart_v1.png"
)

utils.print_html(code_v1, title="LLM output with first draft code")
```

```python
# Get the code within the <execute_python> tags
match = re.search(r"<execute_python>([\s\S]*?)</execute_python>", code_v1)
if match:
    initial_code = match.group(1).strip()
    utils.print_html(initial_code, title="Extracted Code to Execute")
    exec_globals = {"df": df}
    exec(initial_code, exec_globals)

# If code run successfully, the file chart_v1.png should have been generated
utils.print_html(
    content="chart_v1.pn g",
    title="Generated Chart (V1)",
    is_image=True
)
```

```python
# The original code that failed is stored in the 'initial_code' variable.
# We will now build a new prompt to ask the LLM to fix it.

# It's helpful to have the full error message for context.
error_message = "KeyError: 'year'"

# Construct the new, detailed prompt for the LLM
debugging_prompt = f"""
The following Python code, which was intended to create a plot comparing Q1 coffee sales, failed:

--- FAILED CODE ---
{initial_code}
--- END FAILED CODE ---

It produced the following error:
--- ERROR ---
{error_message}
--- END ERROR ---

The error occurred because the DataFrame does not have a 'year' column. All date information is in the 'order_date' column.

Please correct the code to first convert 'order_date' to a datetime object, then extract the year and quarter from the 'order_date' column for filtering. Then, generate the plot again.
"""

# Print the prompt to see what you're sending to the AI
utils.print_html(debugging_prompt, title="New Debugging Prompt")


# Add this to the same cell as the code above

code_v2 = generate_chart_code(
    instruction=debugging_prompt,
    model="openai/gpt-3.5-turbo", # Make sure to use the OpenRouter model name
    out_path_v1="chart_v2.png" # Let's save the new chart with a new name
)

# Display the LLM's new, corrected code output
utils.print_html(code_v2, title="LLM output with corrected code (V2)")
```

```python
# --- CORRECTED CODE BLOCK ---

# Step 1: Extract the corrected code from the LLM's new response (code_v2)
match = re.search(r"<execute_python>([\s\S]*?)</execute_python>", code_v2)

if match:
    # Use a more descriptive variable name for clarity
    corrected_code = match.group(1).strip()
    
    # Print the new code we are about to execute
    utils.print_html(corrected_code, title="Extracted Corrected Code (V2) to Execute")
    
    # Step 2: Execute the corrected code string
    exec_globals = {"df": df}
    exec(corrected_code, exec_globals)
    
    # Step 3: Display the new chart that was just created
    # If the execution was successful, the file chart_v2.png will be displayed
    utils.print_html(
        content="chart_v2.png",  # <-- FIXED: Typo in filename removed
        title="Generated Chart (V2) - Success!",
        is_image=True
    )

else:
    # This will run if the AI didn't return code in the expected format
    print("Could not find code to execute in the LLM's V2 response.")
```
